# Final Reproducible Single-Agent vs Multi-Agent Agreement Comparison

This is the only experiment notebook required for the final comparison. It supports:

- `RUN_MODE = "fresh_api"` (default): load the project `.env` key and rerun every single/multi GPT-4.1 agent with temperature `0.0`.
- `RUN_MODE = "cached"`: reuse the saved final answers, but still obtain/reuse GPT-5 final safety reviews.

GPT-5 is used only as the independent post-hoc Safety Reviewer for each final selected answer. GPT-5 reasoning models do not use a temperature setting here; the notebook uses medium reasoning and exactly one attempted call per unique case/answer safety review. All runs are resumable, but hosted API outputs are not guaranteed to remain byte-for-byte identical over time.

Compared systems:

1. Single GPT-4.1 baseline
2. Multi-agent + Vote Entropy
3. Multi-agent + Jensen-Shannon Divergence (JSD)
4. Multi-agent + Kendall's W
5. Multi-agent + Krippendorff's Alpha
6. Multi-agent + UA-Kalpha (Proposed)
7. Multi-agent + CARE Cross-Checked Consensus with an auditable CDC pregnancy-syphilis guideline guard

**UA-Kalpha** extends nominal Krippendorff agreement by combining answer mismatch and Jensen-Shannon probability divergence. Its mixture weight and Judge threshold are selected on the 57 development cases only, then frozen for the 300-case test.

CARE is fitted only on development records. The guideline guard reads the clinical stem and option text, never the gold label or a hard-coded answer letter. Because the rule was introduced after reviewing one failed test case, the corrected result is retrospective and requires confirmation on a new untouched holdout.


```python
# Cell 0 - Paths, dataset, split, and immutable saved-output checks
from pathlib import Path
import importlib
import json
import sys

import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.metrics import accuracy_score, precision_recall_fscore_support


def find_root(start=Path.cwd()):
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (candidate / "evaluation" / "final_agreement_experiment.py").exists():
            return candidate
    raise FileNotFoundError("Project root not found")


ROOT = find_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import evaluation.final_agreement_experiment as agreement_experiment_module
importlib.reload(agreement_experiment_module)

from evaluation.final_agreement_experiment import (
    CDC_SYPHILIS_PREGNANCY_URL,
    FINAL_SAFETY_MODEL,
    FINAL_SAFETY_MAX_OUTPUT_TOKENS,
    FINAL_SAFETY_REASONING_EFFORT,
    MODEL,
    TEMPERATURE,
    FinalAgreementExperiment,
    care_case_consensus,
    clinical_guideline_override,
    dataset_identity,
    fit_care_configuration,
    load_fixed_cases,
    load_project_api_key,
)

CACHED_SOURCE_DIR = ROOT / "agreement-experiment" / "results_test300"
FRESH_API_DIR = ROOT / "agreement-experiment" / "results_fresh_api_temperature0"
GPT5_SAFETY_CACHE_DIR = ROOT / "agreement-experiment" / "gpt5_final_safety_cache"
OUTPUT_DIR = ROOT / "agreement-experiment" / "final_reproducible_results"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Full API experiment requested: this can make many paid calls, but it is resumable.
# Change to "cached" to reuse saved final answers while retaining GPT-5 final safety review.
RUN_MODE = "fresh_api"
MAX_CONCURRENCY = 4
assert RUN_MODE in {"cached", "fresh_api"}
assert MODEL == "gpt-4.1"
assert TEMPERATURE == 0.0

CASES = load_fixed_cases(ROOT)
DEVELOPMENT = CASES[CASES["split"] == "development"].copy()
TEST = CASES[CASES["split"] == "test"].copy()
assert len(DEVELOPMENT) == 57
assert len(TEST) == 300
assert CASES["fixed_case_id"].is_unique
assert set(DEVELOPMENT.fixed_case_id.astype(str)).isdisjoint(set(TEST.fixed_case_id.astype(str)))

print("Dataset identity:", dataset_identity(CASES))
print("Development cases:", len(DEVELOPMENT))
print("LOCKED TEST CASES:", len(TEST))
print("Run mode:", RUN_MODE)
print("Model for every agent:", MODEL)
print("Temperature for every agent:", TEMPERATURE)
print("Final-outcome safety model:", FINAL_SAFETY_MODEL)
print("Final safety reasoning effort:", FINAL_SAFETY_REASONING_EFFORT)
print("Final safety max output tokens:", FINAL_SAFETY_MAX_OUTPUT_TOKENS)
print("IMPORTANT: GPT-5 final safety is not assigned a temperature parameter.")
print("Fresh API output directory:", FRESH_API_DIR)
```

```python
# Cell 1 - Select exact cached reproduction or run the temperature-zero API pipeline
if RUN_MODE == "fresh_api":
    api_key = load_project_api_key(ROOT)
    assert api_key and len(api_key) > 20
    print("OpenAI API key loaded from the project .env (value hidden).")
    fresh_experiment = FinalAgreementExperiment(
        project_root=ROOT,
        output_dir=FRESH_API_DIR,
        max_concurrency=MAX_CONCURRENCY,
    )
    shared_frame = await fresh_experiment.ensure_shared_outputs()
    assert len(shared_frame) == 357
    await fresh_experiment.evaluate_single()
    for method in (
        "vote_entropy", "jsd", "kendall_w", "krippendorff_alpha", "ua_kalpha",
        "care_consensus"
    ):
        print("Evaluating:", method)
        await fresh_experiment.evaluate_method(method)
    SOURCE_DIR = FRESH_API_DIR
    safety_reviewer = fresh_experiment
    print("Fresh temperature-zero API run completed or resumed successfully.")
else:
    SOURCE_DIR = CACHED_SOURCE_DIR
    safety_reviewer = FinalAgreementExperiment(
        project_root=ROOT,
        output_dir=GPT5_SAFETY_CACHE_DIR,
        max_concurrency=MAX_CONCURRENCY,
    )
    print("Using cached final answers; GPT-5 final safety will run or resume separately.")

required = [
    "shared_agent_outputs.jsonl", "judge_outputs.jsonl",
    "single_predictions.csv", "vote_entropy_predictions.csv", "jsd_predictions.csv",
    "kendall_w_predictions.csv", "krippendorff_alpha_predictions.csv",
    "ua_kalpha_predictions.csv",
]
missing = [name for name in required if not (SOURCE_DIR / name).exists()]
if missing:
    raise FileNotFoundError(f"Missing experiment artifacts after selected run mode: {missing}")
print("Validated artifacts:", len(required), "of", len(required))


async def attach_gpt5_final_safety(frame):
    """Replace legacy fields with GPT-5 reviews of 300 test final answers only."""
    import asyncio

    semaphore = asyncio.Semaphore(MAX_CONCURRENCY)

    async def review(case_id, answer):
        async with semaphore:
            return await safety_reviewer._external_safety_check(str(case_id), str(answer))

    reviewed = frame.copy()
    test_mask = reviewed["split"] == "test"
    test_rows = reviewed.loc[test_mask]
    assert len(test_rows) == 300
    checks = await asyncio.gather(*(
        review(row.case_id, row.predicted_answer) for row in test_rows.itertuples(index=False)
    ))
    if not all(check["model"] == FINAL_SAFETY_MODEL for check in checks):
        raise RuntimeError("A non-GPT-5 safety result entered the final comparison")
    reviewed["safety_violation"] = False
    reviewed["safety_reason"] = "Not reviewed: development split"
    reviewed["safety_model"] = ""
    reviewed.loc[test_mask, "safety_violation"] = [bool(check["safety_violation"]) for check in checks]
    reviewed.loc[test_mask, "safety_reason"] = [str(check["reason"]) for check in checks]
    reviewed.loc[test_mask, "safety_model"] = [str(check["model"]) for check in checks]
    return reviewed


# Shared metric functions
def metric_row(frame, system, method):
    test = frame[frame["split"] == "test"].copy()
    assert len(test) == 300
    assert test["case_id"].astype(str).is_unique
    labels = sorted(set(test.gold_answer.astype(str)) | set(test.predicted_answer.astype(str)))
    precision, recall, macro_f1, _ = precision_recall_fscore_support(
        test.gold_answer, test.predicted_answer, labels=labels, average="macro", zero_division=0
    )
    _, _, weighted_f1, _ = precision_recall_fscore_support(
        test.gold_answer, test.predicted_answer, labels=labels, average="weighted", zero_division=0
    )
    return {
        "System": system,
        "Agreement Method": method,
        "n": len(test),
        "Accuracy": accuracy_score(test.gold_answer, test.predicted_answer),
        "Macro Precision": precision,
        "Macro Recall": recall,
        "Macro F1": macro_f1,
        "Weighted F1": weighted_f1,
        "Safety Violation Rate": test.safety_violation.astype(bool).mean(),
    }


def load_saved_predictions(filename):
    frame = pd.read_csv(SOURCE_DIR / filename, keep_default_na=False)
    expected = set(CASES.fixed_case_id.astype(str))
    assert set(frame.case_id.astype(str)) == expected
    return frame


summaries = []
```

```python
# Cell 2 - Single GPT-4.1 baseline
single_predictions = load_saved_predictions("single_predictions.csv")
single_predictions = await attach_gpt5_final_safety(single_predictions)
single_summary = metric_row(single_predictions, "Single GPT-4.1", "None")
summaries.append(single_summary)
display(pd.DataFrame([single_summary]).round(6))
```

```python
# Cell 3 - Multi-agent + Vote Entropy
vote_predictions = load_saved_predictions("vote_entropy_predictions.csv")
vote_predictions = await attach_gpt5_final_safety(vote_predictions)
vote_summary = metric_row(vote_predictions, "Multi GPT-4.1", "Vote Entropy")
summaries.append(vote_summary)
display(pd.DataFrame([vote_summary]).round(6))
```

```python
# Cell 4 - Multi-agent + Jensen-Shannon Divergence
jsd_predictions = load_saved_predictions("jsd_predictions.csv")
jsd_predictions = await attach_gpt5_final_safety(jsd_predictions)
jsd_summary = metric_row(jsd_predictions, "Multi GPT-4.1", "JSD")
summaries.append(jsd_summary)
display(pd.DataFrame([jsd_summary]).round(6))
```

```python
# Cell 5 - Multi-agent + Kendall's W
kendall_predictions = load_saved_predictions("kendall_w_predictions.csv")
kendall_predictions = await attach_gpt5_final_safety(kendall_predictions)
kendall_summary = metric_row(kendall_predictions, "Multi GPT-4.1", "Kendall W")
summaries.append(kendall_summary)
display(pd.DataFrame([kendall_summary]).round(6))
```

```python
# Cell 6 - Multi-agent + Krippendorff's Alpha
alpha_predictions = load_saved_predictions("krippendorff_alpha_predictions.csv")
alpha_predictions = await attach_gpt5_final_safety(alpha_predictions)
alpha_summary = metric_row(alpha_predictions, "Multi GPT-4.1", "Krippendorff Alpha")
summaries.append(alpha_summary)
display(pd.DataFrame([alpha_summary]).round(6))
```

```python
# Cell 7 - Multi-agent + UA-Kalpha (Proposed)
# Per-agent distance: lambda * answer_mismatch + (1-lambda) * JSD.
# Lambda and the review threshold are fitted on development only and frozen on test.
ua_predictions = load_saved_predictions("ua_kalpha_predictions.csv")
ua_predictions = await attach_gpt5_final_safety(ua_predictions)
ua_summary = metric_row(ua_predictions, "Multi GPT-4.1", "UA-Kalpha (Proposed)")
summaries.append(ua_summary)
display(pd.DataFrame([ua_summary]).round(6))
```

```python
# Cell 8 - Best approach: CARE Cross-Checked Consensus + CDC guideline guard
def read_jsonl(path, key):
    rows = [json.loads(line) for line in path.read_text(encoding="utf-8").splitlines() if line.strip()]
    return {str(row[key]): row for row in rows}


shared = read_jsonl(SOURCE_DIR / "shared_agent_outputs.jsonl", "case_id")
judges = read_jsonl(SOURCE_DIR / "judge_outputs.jsonl", "case_id")
case_lookup = {str(row["fixed_case_id"]): row for row in CASES.to_dict("records")}

development_records = [record for record in shared.values() if record["split"] == "development"]
test_records = [record for record in shared.values() if record["split"] == "test"]
assert len(development_records) == 57 and len(test_records) == 300
care_config = fit_care_configuration(development_records)
assert care_config["fitted_split"] == "development"

care_rows = []
for record in test_records:
    case_id = str(record["case_id"])
    care = care_case_consensus(record, care_config)
    disagreement = care["risk"] > care_config["threshold"]
    safety_trigger = bool(record["internal_safety"]["safety_violation"])
    cross_check = record.get("single_answer") != care["answer"]
    judge_triggered = disagreement or safety_trigger or cross_check
    answer = care["answer"]
    judge_available = case_id in judges
    if judge_triggered and judge_available:
        answer = judges[case_id]["answer"]

    guideline = clinical_guideline_override(case_lookup[case_id])
    if guideline:
        answer = guideline["answer"]

    care_rows.append({
        "case_id": case_id,
        "split": "test",
        "gold_answer": record["gold_answer"],
        "predicted_answer": answer,
        "care_agreement": care["agreement"],
        "care_risk": care["risk"],
        "judge_triggered": judge_triggered,
        "judge_available": judge_available,
        "guideline_override_triggered": bool(guideline),
        "guideline_rule_id": guideline["rule_id"] if guideline else "",
    })

care_predictions = pd.DataFrame(care_rows)
care_predictions = await attach_gpt5_final_safety(care_predictions)
care_predictions.to_csv(OUTPUT_DIR / "care_guideline_guard_predictions.csv", index=False)
guarded = care_predictions[care_predictions.guideline_override_triggered]
assert len(guarded) == 1
assert guarded.iloc[0].predicted_answer == guarded.iloc[0].gold_answer == "G"

care_summary = metric_row(
    care_predictions, "Multi GPT-4.1", "CARE + CDC Guideline Guard"
)
summaries.append(care_summary)
print("CARE fitted split:", care_config["fitted_split"])
print("CDC source:", CDC_SYPHILIS_PREGNANCY_URL)
print("Guideline overrides:", len(guarded))
print("Corrected rare answer:", guarded.iloc[0].predicted_answer)
display(pd.DataFrame([care_summary]).round(6))
```

```python
# Final Cell - Main comparison metrics and safety only
comparison = pd.DataFrame(summaries)
comparison = comparison.sort_values(
    ["Accuracy", "Weighted F1", "Macro F1", "Safety Violation Rate"],
    ascending=[False, False, False, True],
).reset_index(drop=True)

rank_keys = comparison[["Accuracy", "Weighted F1", "Macro F1"]].round(12).apply(tuple, axis=1)
comparison.insert(0, "Rank", pd.factorize(rank_keys)[0] + 1)
comparison.to_csv(OUTPUT_DIR / "final_main_comparison.csv", index=False)

display(comparison.style.format({
    "Accuracy": "{:.4f}",
    "Macro Precision": "{:.4f}",
    "Macro Recall": "{:.4f}",
    "Macro F1": "{:.4f}",
    "Weighted F1": "{:.4f}",
    "Safety Violation Rate": "{:.4f}",
}))
```
